# Correlation

```python
import seaborn as sns

corr = df.corr(numeric_only=True)
matrix = np.triu(np.ones_like(corr))
ax = sns.heatmap(corr, annot=True, mask=matrix)
```

```python
ax = sns.heatmap(corr, annot=True, mask=(corr < 0) | matrix)
```
